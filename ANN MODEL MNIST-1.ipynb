Struct of ANN:

#Model building


```python
x_train_scale.shape
```

```python
x_test_scale.shape
```

```python
#import tensorflow as tf # Tensorflow used for building and training neural networks
from tensorflow.keras.models import Sequential # Sequential model
from tensorflow.keras.layers import Input, Flatten, Dense   # Input Layer, Flatten Layer, Dense Layer

```

```python
ann_model = Sequential()


ann_model.add(Flatten(input_shape=(28,28))) # This will convert 2D array to 1D array and then feed to input layer

ann_model.add(Dense(units=128, activation='relu'))# First hidden layer with 128 neurons and ReLU activation function is added to the model
ann_model.add(Dense(units=64, activation='relu'))# Second hidden layer with 64 neurons and ReLU activation function is added to the model
ann_model.add(Dense(units=32, activation='relu'))# Third hidden layer with 32 neurons and ReLU activation function is added to the model
ann_model.add(Dense(units=10, activation='softmax'))# Output layer with 10 neurons (for 10 classes) and Softmax activation function is added to the model

#Summary of the model
ann_model.summary()
```

Add compile with Optimizer, accuracy and loss

```python

from tensorflow.keras.optimizers import Adam

ann_model.compile(
    optimizer=Adam(learning_rate=0.001), #Optimizer
              loss='sparse_categorical_crossentropy',# Loss funtion for multi-class classification
              metrics=['accuracy']
              )
```

```python
#We added input layers, hidden layers, output layer and compiled the model with optimizer, loss function and accuracy metric.
```

Training the model with data

```python
ann_model.fit(x_train_scale, y_train, epochs=10,batch_size=100) 
```

# Evaluate on the test set

```python
test_loss, test_acc= ann_model.evaluate(x_test_scale, y_test,verbose=2)
print(f"Test loss: {test_loss*100}, Test accuracy: {test_acc*100}")
```

Plot history (loss & accuracy)

```python
import matplotlib.pyplot as plt
 #accuracy plot
history = ann_model.fit(x_train_scale, y_train, epochs=10, batch_size=100, validation_data=(x_test_scale, y_test))

plt.figure(figsize=(8,4))
plt.plot(history.history['accuracy'], label='train_acc')
plt.plot(history.history['val_accuracy'], label='val_acc')
plt.title('Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

```

```python
#loss plot
plt.figure(figsize=(8,4))
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.title('Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

Confusion matrix & sample misclassifications

```python
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np
import matplotlib.pyplot as plt

y_pred = ann_model.predict(x_test_scale)
y_pred_labels = np.argmax(y_pred, axis=1)
print(classification_report(y_test, y_pred_labels))

cm = confusion_matrix(y_test, y_pred_labels)
print('confusion matrix:\n', cm)

# show some misclassification examples
mis_idx = np.where(y_pred_labels != y_test)[0][:12]
plt.figure(figsize=(10, 6))

for i, idx in enumerate(mis_idx):
    plt.subplot(3, 4, i + 1)
    plt.imshow(x_test_scale[idx])
    plt.title(f'True:{y_test[idx]}, Pred:{y_pred_labels[idx]}')
    plt.axis('off')
plt.tight_layout()
plt.show()

```
